# Tuple, Listen, Dictionaries und Indexe und Schlüssel

Tuple, Listen und Dictionaires sind Strukturen zum speichern mehrerer Werte.

Indexe und Schlüssel werden verwendet um aus ihnen die einzelnen werte ab zu speichern.

## [ Listen ]

```python
# listen definiert man mit Eckicken klammern [[[[[[[[[[[[[
# bei einer definition wird der wert ohne variablen name erstellt

[1, 2, 3, 4] # das ist eine feiliegende listen definition die nirgendwo gespeichert wird

# irgend_ein_name[1,2,3,4] # das ist kein gültiger python code!

# um den den aber verwenden zu können müssen wir ihn erst einer Variable zuweisen
# dazu wird der variablen name angegeben und der wert wird mit hilfe des zuweisungsoperators (=)
# hier ein beispiel:

meine_tolle_liste = [11, 22, 33, 44, 55, 66, 77, 88, 99]
```

## { Dictionaries }

```python
# Dictionaries (zu deutsch Wörterbücher) definiert man mit geschweiften Klammern
# wie bei jeder werte-definition darf hier kein variablen name direkt angrenzen
# in einem wörterbuch wird ein wert immer einem schlüssel zugewiesen dazu wird ein doppelpunkt verwendet

{"A":1, "B":2, "C":3} # das ist einer freiliegende wörterbuch defininition die nicht gespeichert wird

# irgend_ein_name{"A":1, "B":2, "C":3} # das ist kein gültiger python code!

# zuweisung sind immer gleich in python also genau wie bei der liste der variablen name gefolgt vom zuweisungs operator und einem wert

mein_tolles_dict = {"A":1, "B":2, "C":3}
```

## Abfrage von werten aus dictionaries oder listen

*geht auch bei tuplen*

Um aus einer Liste oder einem Wörterbuch die Werte direkt ab fragen zu können benutzt man den **Index-Operator**.

Dieser wird verwendet indem man den Index oder den Schlüssel in Eckigen-Klammern hinter der liste (entweder variablen-name oder wert) angibt.

Hier ein Paar Beispiele

### listen wert abfrage

um aus einer liste den ersten wert zu bekommen müssen wir den ersten index abfragen
indexe in python beginnen bei 0
speziel in python sind auch negative indexe möglich. diese greifen von ende der liste zu.
die formel für den ziel index lautet hierbei ``listen-länge + negativ-index``
damit bekommt man zB. mit dem negativen index -1 den letzten wert einer liste.

#### rechnungs beispiel
bei einer liste mit fünf werten ``["a", "b", "c", "d", "e"]`` würde sich das also so verhalten dass:
die länge ist ``5`` dann plus ``-1`` heißt ``5 + -1`` das ergibt ``4``. bei einer liste mit einer länge von ``5`` ist der
letzte index ``4`` und somit bekommen wir den letzten wert auch mit dem index ``-1``
bei index ``-2`` wäre es der vorletzte wert und
bei index ``-3`` wäre es somit der vorvorletzte wert

bei einer liste kann ich mit dem index operator auch einen bereich der liste abfragen. das ergibt dann eine neue liste.
hierbei muss ich zwei indexe übergeben die ich mit einem doppelpunk ``:`` trenne.
wenn ich aber den doppelpunk angebe aber einen der werte weg lasse, geht die liste ohne bedingung so weit wie es geht. (*1)
ich kan hierbei auch die schritte für den listenbereich angeben. sies tuhe ich mit einem weiteren doppelpunkt und einem weiteren wert. ist dieser nicht angegeben wird standartmässig mit schritten von ``1`` gearbeitet.
WICHTIG: Der zweite index ist die grenze des Bereiches und nicht inklusive.

#### rechnungs beispiel
wenn ich einen schritt von ``2`` angebe bekomme ich nur jeden zweiten wert, mit einem schritt von 3 immer nur jeden dritten usw. und mit einem schritt von -1 würde ich rückwärtz durch die liste laufen (``Beispiel: "Hallo Welt!"[::-1] wäre dann "!tleW ollaH"``) und mit einem schritt von -2 rückwärtz aber nur jeden zweiten.


```python
# abfrage eines einzelnen index hier mal der wert von index 3 gespeichert in der variable n

n = meine_tolle_liste[3]

# abfrage eines bereiches ab dem start der liste bis hin aber nicht inklusive der letzten beiden werte

b = meine_tolle_liste[:-2]

# abfrage jedes dritten wertes beginnend bei dem zweiten wert der liste und bis nicht inclusive der letzten 3 werte

wtf = meine_tolle_liste[1:-3:3]
```
